## 1. Description

In the fourth and last DAML project we will explore whether a ML classifier can improve a traditional search for exotic particles with ATLAS data. The project builds upon the Semester-2 Week-9 checkpoint.

We evaluate the statistical significance of a signal-like deviation in simulated data using two approaches:

- The **traditional approach**: apply "square" selection cuts on kinematic variables, then perform two fits (for the $H_0$ and $H_1$ hypotheses) on the mass spectrum, and use the fit results to quantify the size of the deviation.
- The **ML approach**: incorporate a ML classifier to improve the $S/B$ ratio in the region of interest, followed by the same two-fit procedure and quantitative evaluation of the deviation.

The goal is to determine if the ML classifier can improve the sensitivity of the search, and quantify any gains.

## 2. Input Files

CSV files with kinematic information for approximately 6.8M background and 50k signal simulated collisions are available at: https://cernbox.cern.ch/s/WLHOLZEsxUQiSeo

The background events are split into three files corresponding to different physics processes. The signal sample has been generated for a hypothetical heavy Higgs boson with $m_H = 1\ \text{TeV}/c^2$. Each entry corresponds to a separate collision event. The relevant variables are summarised below:

| `lep1_pt` | Transverse momentum of first lepton (in $\text{MeV}/c$) |
| `lep2_pt` | Transverse momentum of second lepton (in $\text{MeV}/c$) |

## 4. Fitting and Hypothesis Testing — *[30%]*

- Model the `reco_zv_mass` spectrum for signal and background **separately** after selection cuts, using appropriate fitting functions (e.g. a single Gaussian for signal, a polynomial for background). Perform the fit over a subrange of the mass spectrum.

- Model the joint **signal-plus-background** spectrum as a weighted sum of the two fit functions. Fix the signal shape parameters (e.g. mean and $\sigma$ of the Gaussian) to the values from the signal-only fit, but leave the overall signal normalisation free. Treat all background shape parameters as free.

- Carry out two fits under the $H_1$ (signal + background) and $H_0$ (background only) hypotheses. Apply **Wilks' theorem** to calculate the statistical significance of the signal-like deviation.

 **Important**: Because we use weighted events, we need a modified $\chi^2$ for minimisation. For a binned fit, the observed count and variance in bin $i$ are adjusted as follows:

 $$N^{(i)}_\text{observed} \;\longrightarrow\; \sum_{j=1}^{n} w^{(i)}_j$$

 $$\sigma^2_i \;\longrightarrow\; \sum_{j=1}^{n} \left(w^{(i)}_j\right)^2$$

 giving a modified $\chi^2$ of:

 $$\chi^2 = \sum_i \frac{\left[\displaystyle\sum_{j=1}^{n} w^{(i)}_j - N^{(i)}_\text{predicted}\right]^2}{\displaystyle\sum_{j=1}^{n} \left(w^{(i)}_j\right)^2}$$
